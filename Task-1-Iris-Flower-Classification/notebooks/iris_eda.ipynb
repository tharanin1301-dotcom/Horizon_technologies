# Iris Dataset – Exploratory Data Analysis
This notebook demonstrates a quick EDA on the Iris dataset using Pandas, Seaborn and Matplotlib. It mirrors the steps performed in the main pipeline script and can be extended with additional analysis.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load data
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=["sepal_length", "sepal_width", "petal_length", "petal_width"])
df["species"] = pd.Categorical.from_codes(iris.target, iris.target_names)
df.head()
```

```python
# Class distribution
sns.countplot(x='species', data=df, palette='viridis')
plt.title('Iris Species Distribution')
plt.show()
```

```python
# Pairplot (optional)
sns.pairplot(df, hue='species', palette='bright')
plt.suptitle('Pairplot of Iris Measurements', y=1.02)
plt.show()
```
